````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (2) 建立資料庫與初始資料
import sqlite3

conn = sqlite3.connect('activity_system.db')
cursor = conn.cursor()

try:
    cursor.execute('DROP TABLE IF EXISTS registrations;')
    cursor.execute('DROP TABLE IF EXISTS activities;')

    # 建立活動表 (activities)，包含日期與名額
    cursor.execute('''
        CREATE TABLE activities (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            date TEXT NOT NULL,
            max_capacity INTEGER NOT NULL
        );
    ''')

    # 建立報名紀錄表 (registrations)，包含狀態欄位
    cursor.execute('''
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            user_name TEXT NOT NULL,
            activity_name TEXT NOT NULL,
            count INTEGER NOT NULL,
            status TEXT NOT NULL
        );
    ''')

    # 放入指定的初始資料
    initial_activities = [
        ('迎新晚會', '2026-11-20', 6),
        ('程式工作坊', '2026-11-25', 4),
        ('籃球友誼賽', '2026-11-28', 3)
    ]
    cursor.executemany('INSERT INTO activities (name, date, max_capacity) VALUES (?, ?, ?);', initial_activities)

    conn.commit()
    print("✅ 資料庫與初始資料建立成功！")
except Exception as e:
    conn.rollback()
    print(f"❌ 建立資料庫時發生錯誤：{e}")
finally:
    conn.close()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (4) AI 解讀函式
import time
import re

ai_cache = {}

def parse_intent(text):
    if text in ai_cache:
        return ai_cache[text]

    prompt = f"""
    你是一個系學會活動報名系統的語意解析助理。
    請從使用者的輸入中解析出三個欄位：動作、活動、人數。
    格式固定為：動作,活動,人數

    - 動作只能是以下其中之一：
      - 報名
      - 取消
      - 查詢
      - 統計
      - UNKNOWN
    - 活動只能是以下其中之一：
      - 迎新晚會
      - 程式工作坊
      - 籃球友誼賽
      - NONE
    - 人數：請抓出數字（若提到「我和兩個朋友」代表 3 人；若未提及且為報名/取消則預設為 1；查詢或統計填 0）。

    請嚴格只輸出「動作,活動,人數」格式，不要有多餘文字。
    使用者輸入：{text}
    """

    response_text = ""
    for attempt in range(3):
        try:
            kwargs = {"model": MODEL, "temperature": 0}
            if 'REASONING_EFFORT' in globals() and REASONING_EFFORT:
                kwargs["reasoning_effort"] = REASONING_EFFORT

            completion = client.chat.completions.create(
                messages=[{"role": "user", "content": prompt}],
                **kwargs
            )
            response_text = completion.choices[0].message.content.strip()
            time.sleep(2)
            break
        except Exception as e:
            if "429" in str(e) or "rate_limit" in str(e).lower():
                time.sleep(30)
            else:
                break

    if not response_text:
        response_text = "UNKNOWN,NONE,0"

    allowed_actions = ["報名", "取消", "查詢", "統計", "UNKNOWN"]
    allowed_activities = ["迎新晚會", "程式工作坊", "籃球友誼賽", "NONE"]

    parts = [p.strip() for p in response_text.split(",")]

    action = parts[0] if len(parts) > 0 and parts[0] in allowed_actions else "UNKNOWN"
    activity = parts[1] if len(parts) > 1 and parts[1] in allowed_activities else "NONE"

    count = 1
    if len(parts) > 2:
        match_num = re.search(r'\d+', parts[2])
        if match_num:
            count = int(match_num.group())

    if action in ["查詢", "統計"]:
        count = 0

    result = (action, activity, count)
    ai_cache[text] = result
    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (5) AI 解讀測試
test_cases = [
    ("我要報名迎新晚會 2 個人", "報名", "迎新晚會", 2),
    ("我和兩個朋友參加程式工作坊", "報名", "程式工作坊", 3),
    ("請幫我取消籃球友誼賽", "取消", "籃球友誼賽", 1),
    ("請問程式工作坊還有名額嗎？", "查詢", "程式工作坊", 0),
    ("各活動報名狀況", "統計", "NONE", 0),
    ("報名迎新晚會 4 個人", "報名", "迎新晚會", 4),
    ("我要報名籃球友誼賽", "報名", "籃球友誼賽", 1),
    ("明天天氣怎麼樣？", "UNKNOWN", "NONE", 0),
    ("學校有幾棟大樓？", "UNKNOWN", "NONE", 0)
]

passed_count = 0
print("=== 🧪 AI 解讀測試開始 ===")

for i, (text, expected_action, expected_activity, expected_count) in enumerate(test_cases, 1):
    parsed_action, parsed_activity, parsed_count = parse_intent(text)

    action_match = (expected_action == "*" or parsed_action == expected_action)
    activity_match = (expected_activity == "*" or parsed_activity == expected_activity)
    count_match = (expected_count == "*" or parsed_count == expected_count)

    is_passed = action_match and activity_match and count_match
    if is_passed:
        passed_count += 1
        status = "✅ 通過"
    else:
        status = "❌ 失敗"

    print(f"\n測資 {i}: 「{text}」")
    print(f"  -> 預期: 動作={expected_action}, 活動={expected_activity}, 數量={expected_count}")
    print(f"  -> 系統解讀: 動作={parsed_action}, 活動={parsed_activity}, 數量={parsed_count} [{status}]")

print(f"\n========================")
print(f"🎯 測試結果：總共 {len(test_cases)} 句，成功通過 {passed_count} 句。")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (6) 功能函式
import sqlite3

def get_connection():
    return sqlite3.connect('activity_system.db')

def register_activity(user_name, activity_name, count):
    conn = get_connection()
    cursor = conn.cursor()
    try:
        # A1 需求：人數必須介於 1 到 3 之間
        if not (1 <= count <= 3):
            return f"報名失敗：每筆報名人數必須介於 1 到 3 之間（您輸入了 {count} 人）。"

        # 檢查活動是否存在
        cursor.execute("SELECT max_capacity FROM activities WHERE name = ?;", (activity_name,))
        row = cursor.fetchone()
        if row is None:
            return f"報名失敗：找不到活動「{activity_name}」。"
        max_cap = row[0]

        # A2 需求：檢查該學生目前已同時報名幾個活動（status = '已報名'）
        cursor.execute("""
            SELECT COUNT(DISTINCT activity_name) FROM registrations
            WHERE user_name = ? AND status = '已報名';
        """, (user_name,))
        active_courses_count = cursor.fetchone()[0]

        # 檢查是否重複報名同一個活動
        cursor.execute("""
            SELECT COUNT(*) FROM registrations
            WHERE user_name = ? AND activity_name = ? AND status = '已報名';
        """, (user_name, activity_name))
        already_registered = cursor.fetchone()[0] > 0

        if already_registered:
            return f"報名失敗：您已經報名過「{activity_name}」了，不可重複報名。"

        if active_courses_count >= 2:
            return f"報名失敗：每位學生最多同時報名 2 個活動（您目前已報名 2 個）。"

        # 檢查剩餘名額
        cursor.execute("SELECT SUM(count) FROM registrations WHERE activity_name = ? AND status = '已報名';", (activity_name,))
        sum_row = cursor.fetchone()
        current_booked = sum_row[0] if sum_row and sum_row[0] is not None else 0
        remaining = max_cap - current_booked

        if remaining < count:
            return f"報名失敗，「{activity_name}」剩餘名額不足！目前僅剩 {remaining} 人。"

        # 寫入資料庫
        cursor.execute(
            "INSERT INTO registrations (user_name, activity_name, count, status) VALUES (?, ?, ?, '已報名');",
            (user_name, activity_name, count)
        )
        conn.commit()
        return f"成功報名「{activity_name}」{count} 人！"
    except Exception as e:
        conn.rollback()
        return f"報名發生錯誤：{e}"
    finally:
        conn.close()

def cancel_activity(user_name, activity_name, count):
    conn = get_connection()
    cursor = conn.cursor()
    try:
        cursor.execute(
            "SELECT id FROM registrations WHERE user_name = ? AND activity_name = ? AND status = '已報名';",
            (user_name, activity_name)
        )
        row = cursor.fetchone()
        if row is None:
            return f"取消失敗：找不到您（{user_name}）在「{activity_name}」的有效報名紀錄。"

        reg_id = row[0]
        # 依照規格：取消時把紀錄狀態改為『已報名』->『已取消』並釋出名額
        cursor.execute("UPDATE registrations SET status = '已取消' WHERE id = ?;", (reg_id,))
        conn.commit()
        return f"已成功取消「{activity_name}」的報名，名額已釋出。"
    except Exception as e:
        conn.rollback()
        return f"取消發生錯誤：{e}"
    finally:
        conn.close()

def query_activity(activity_name):
    conn = get_connection()
    cursor = conn.cursor()
    try:
        cursor.execute("SELECT date, max_capacity FROM activities WHERE name = ?;", (activity_name,))
        row = cursor.fetchone()
        if row is None:
            return f"查詢失敗：找不到活動「{activity_name}」。"
        date_str, max_cap = row

        cursor.execute("SELECT SUM(count) FROM registrations WHERE activity_name = ? AND status = '已報名';", (activity_name,))
        sum_row = cursor.fetchone()
        current_booked = sum_row[0] if sum_row and sum_row[0] is not None else 0
        remaining = max_cap - current_booked

        return f"「{activity_name}」於 {date_str} 舉辦，總名額 {max_cap} 人，已報名 {current_booked} 人，目前剩餘名額為 {remaining} 人。"
    except Exception as e:
        return f"查詢發生錯誤：{e}"
    finally:
        conn.close()

def get_statistics():
    # A3 需求：使用 JOIN 與 GROUP BY 列出每個活動的日期、名額、已報名人數、剩餘額。沒有人報名的活動也必須列出。
    conn = get_connection()
    cursor = conn.cursor()
    try:
        query = """
            SELECT
                a.name,
                a.date,
                a.max_capacity,
                COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.count ELSE 0 END), 0) AS total_booked
            FROM activities a
            LEFT JOIN registrations r ON a.name = r.activity_name
            GROUP BY a.id, a.name, a.date, a.max_capacity;
        """
        cursor.execute(query)
        rows = cursor.fetchall()

        result_msg = "📊【各活動報名統計報表】\n"
        for row in rows:
            name, date_str, max_cap, booked = row
            remaining = max_cap - booked
            result_msg += f"- {name} ({date_str}): 總名額 {max_cap} 人 | 已報名 {booked} 人 | 剩餘 {remaining} 人\n"
        return result_msg.strip()
    except Exception as e:
        return f"統計查詢發生錯誤：{e}"
    finally:
        conn.close()

def handle_user_input(text, user_name="小明"):
    action, activity, count = parse_intent(text)

    if action == "報名":
        return register_activity(user_name, activity, count)
    elif action == "取消":
        return cancel_activity(user_name, activity, count)
    elif action == "查詢":
        return query_activity(activity)
    elif action == "統計":
        return get_statistics()
    else:
        return "無法辨識您的意圖。目前的活動有：迎新晚會、程式工作坊、籃球友誼賽。"
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (7) 主流程與流程測試
import sqlite3
import pandas as pd
from IPython.display import display

conn = sqlite3.connect('activity_system.db')
cursor = conn.cursor()
cursor.execute('DELETE FROM registrations;')
conn.commit()
conn.close()

workflow_test_steps = [
    (1, "小明", "請問迎新晚會還有名額嗎？"),
    (2, "小明", "我要報名迎新晚會 2 個人"),
    (3, "小明", "我和兩個朋友參加程式工作坊"),
    (4, "小明", "我要報名籃球友誼賽"),
    (5, "小華", "我要報名迎新晚會 5 個人"),
    (6, "小明", "請幫我取消迎新晚會"),
    (7, "小明", "我要報名籃球友誼賽"),
    (8, "幹部", "各活動報名狀況"),
    (9, "小美", "明天會下雨嗎？")
]

print("=== 🚀 系統主流程與 9 步流程測試開始 ===\n")

for step_num, name, text in workflow_test_steps:
    print(f"步驟 {step_num} [{name}] 說：「{text}」")
    response = handle_user_input(text, user_name=name)
    print(f"  👉 系統回應：{response}\n")

print("=== 最終資料表狀態 ===")
conn = sqlite3.connect('activity_system.db')
try:
    display(pd.read_sql_query("SELECT * FROM registrations;", conn))
finally:
    conn.close()
````
